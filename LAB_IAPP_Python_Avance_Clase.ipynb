```python
#3.
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]


print(f"1. El nombre del primer producto es: {inventario[0]['nombre']}")

inventario.append({"nombre": "Huevos", "precio": 6000, "stock": 20})
print("2. Producto agregado. Inventario actual:", inventario[3])

for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
print(f"3. Stock de Leche actualizado: {inventario[1]['stock']} unidades")

valor_total = sum(p["precio"] * p["stock"] for p in inventario)
print(f"4. El valor total del inventario es: {valor_total}")

print("5. Productos con stock menor a 10:")
menores_a_10 = [p for p in inventario if p["stock"] < 10]
if menores_a_10:
    for p in menores_a_10:
        print(f"   - {p['nombre']} (Stock: {p['stock']})")
else:
    print("   No hay productos con stock menor a 10.")

for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)
print("6 y 7. Precios aumentados un 10% y redondeados:", inventario)

nombre_buscado = "Café"
encontrado = False

for producto in inventario:
    if not encontrado:
      if producto["nombre"].lower()==nombre_buscado.lower():
        print(f"8. producto encontrado {producto}")
        encontrado = True
if not encontrado:
  print(f"8. El producto "{nombre_buscado}" no se encuentra en el inventario.")
inventario[0]["stock"] = 0

inventario = [p for p in inventario if p["stock"] > 0]
print("9. Inventario tras eliminar productos con stock 0:", inventario)

inventario.sort(key=lambda x: x["precio"])
print("10. Inventario ordenado por precio ascendente:")
for p in inventario:
    print(f"    - {p['nombre']}: ${p['precio']} (Stock: {p['stock']})")
```

```python
#5.

productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))


    if precio < 0 or cantidad < 0:
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)

if productos:
    mas_caro = productos[0]
    for p in productos:
        if p["precio"] > mas_caro["precio"]:
            mas_caro = p
    print("Producto más caro:", mas_caro["nombre"])

print(productos)


```

```python
# 1.
import requests

# Llamado a una API pública
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()
# Procesamiento imperativo
nombres = []
for u in usuarios:
    if u["email"]. endswith(".biz"):
      nombres.append(u["name"])

print("Usuarios encontrados cuyo correo termina en .biz:")
for n in nombres:
    print("-", n)
```
